```python
import pandas as pd 
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python



transactions_info =    {
        "table":"frauddetection.bronze.transactions",
        "file":"/transactions/transactions_500k.csv"
    }


pd_df = pd.read_csv(
    f"https://frauddetectionabcbank.blob.core.windows.net/bronze{transactions_info['file']}?sp=r&st=2026-03-25T03:25:25Z&se=2026-10-15T11:40:25Z&spr=https&sv=2024-11-04&sr=c&sig=xaoFm%2BaKYt%2BHY2ENUJZn3uIbDq8ctRLQusX9VBsRxPI%3D"
)


df = spark.createDataFrame(pd_df)\
    .withColumn("ingestTimeStamp",current_timestamp())\
        .withColumn("ingestFileName",lit(transactions_info["file"]))

df.write.format("delta")\
    .mode("overwrite")\
        .option("mergeSchema","true")\
        .option("delta.enableChangeDataFeed","true")\
            .saveAsTable(f"{transactions_info["table"]}")



```
